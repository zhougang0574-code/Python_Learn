### 4、*args 和 **kwargs

```python
# 1. 写函数 my_sum(*args)，返回所有传入数字的和，调用 my_sum(1, 2, 3, 4, 5)
# 2. 写函数 build_info(**kwargs)，拼接 "key=value" 字符串，调用 build_info(name="Tom", age=18, city="Beijing")
# 3. 用 * 拆包列表 [1,2,3] 传给 my_sum，打印结果
# 4. 用 ** 拆包字典 {"name":"Tom","age":18} 传给 build_info，打印结果
# 期望输出：
# 15
# name=Tom, age=18, city=Beijing
# 6
# name=Tom, age=18

# 在这里写你的代码

def my_sum(*args):
    return sum(args)


def build_info(**kwargs):
    return "，".join(f"{k}={v}" for k, v in kwargs.items())


build_info(name="Tom", age=18, city="Beijing")
nums = [1, 2, 3]
my_sum(*nums)
kv = {"name": "Tom", "age": 18}
print(build_info(**kv))
```

### 5、变量作用域

```python
# 1. 写全局变量 counter = 0，写函数 click() 用 global 让 counter 加 1，调用 3 次后打印 counter
# 2. 写嵌套函数 outer()/inner()，用 nonlocal 让 inner 修改 outer 的 x，调用 counter() 三次
# 期望输出：
# 3
# 1 2 3

# 在这里写你的代码
counter = 0


def click():
    global counter
    for i in range(0, 3):
        counter += 1


click()
print(counter)


def outer():
    x = 0

    def inner():
        nonlocal x
        x += 1
        return x

    return inner


co = outer()
print(co(), co(), co())
```

### 7、函数是一等对象

```python
# 1. 写函数 apply_twice(func, x)，对 x 应用 func 两次
#    测试：apply_twice(lambda x: x*2, 3) 打印结果
# 2. 写函数 make_multiplier(n)，返回一个函数，把输入乘以 n
#    测试：double = make_multiplier(2)，打印 double(5)
# 3. 把函数存入列表 [len, str.lower, str.upper]，分别调用处理 "Hello"
# 4. 把函数存入字典 {"add": lambda a,b: a+b, "mul": lambda a,b: a*b}，调用 ops["add"](3, 4)
# 5. 写函数 choose(op)，根据 op 返回 add 或 mul 函数，调用 choose("mul")(3, 4)
# 期望输出：
# 12
# 10
# 5 hello HELLO
# 7
# 12

# 在这里写你的代码
def apply_twice(func, x):
    return func(func(x))
print(apply_twice(lambda x: x*2 ,3))
def make_multiplier(n):
    def mult(x):
        return x * n
    return mult
double = make_multiplier(2)
print(double(5))

ls = [len, str.lower, str.upper]
for i in ls:
    print(i("Hello"))

ops = {"add": lambda a,b: a+b, "mul": lambda a,b: a*b}
print(ops["add"](3,4))

def add(a,b):
    return a + b

def mul(a,b):
    return a*b
def choose(op):
    if op == "add":
        return add
    elif op =="mul":
        return mul
print(choose("add")(3,4))
```

```python
# 1. 写一个装饰器 @log_call，让被装饰函数每次调用时打印：
#    "[LOG] 调用 xxx，参数: (...)"
#    测试：
#    @log_call
#    def add(a, b):
#        return a + b
#    print(add(3, 4))
#    期望：先打印 [LOG] 调用 add，参数: (3, 4)，再打印 7
def log_call(func):
    def wrapper(*args,**kwargs):
        print(f"[LOG] 调用 {func.__name__}, 参数 ({args})")
        result = func(*args,**kwargs)
        return result
    return wrapper
@log_call
def add(a, b):
    return a + b
print(add(3, 4))

# 2. 写一个装饰器 @timer，测量函数运行时间并打印
#    测试：
#    @timer
#    def slow_add(a, b):
#        time.sleep(0.3)
#        return a + b
#    print(slow_add(3, 4))
#    期望：打印 slow_add 耗时: 0.3xxx s，再打印 7
# 3. 写一个装饰器 @double_result，将函数的返回值乘以 2
#    测试：
#    @double_result
#    def add(a, b):
#        return a + b
#    print(add(3, 4))
#    期望：14（因为 3+4=7，翻倍后=14）
#
# 4. 写一个装饰器 @bold，将返回的字符串用 <b>...</b> 包裹
#    测试：
#    @bold
#    def greet(name):
#        return f"你好, {name}!"
#    print(greet("Tom"))
#    期望：<b>你好, Tom!</b>

# 在这里写你的代码
```
