### 1、if / elif / else 条件分支

```python
# 写一个判断：输入分数 score = 73
# >=90 是 "A"，>=80 是 "B"，>=60 是 "C"，否则是 "D"，打印等级
# 期望输出：C

# 在这里写你的代码
def score(score:int):
    if score >= 90:
        print("A")
    elif score >= 80:
        print("B")
    elif score >= 60:
        print("C")

score(73)
```

### 3、三元表达式

```python
# 1. 用三元表达式判断 n=7 是 "偶数" 还是 "奇数"
# 2. 用嵌套三元表达式判断 x=5 是 "正"、"零" 还是 "负"
# 期望输出：
# 奇数
# 正

# 在这里写你的代码
n = 7
print("偶数" if n%2==0 else "奇数")
x = 5
print("零" if x == 0 else ("负" if x < 0 else "正"))
```

### 4、for 循环：range / enumerate / zip

```python
# 1. 用 range(5) 打印 0 1 2 3 4
# 2. 用 range(2, 10, 2) 打印 2 4 6 8
# 3. 用 enumerate 遍历 fruits=["apple", "banana", "cherry"]，打印 "0: apple" 等
# 4. 用 zip 同时遍历 names=["Tom", "Jerry"] 和 ages=[18, 20]，打印 "Tom 18" 等
# 期望输出：
# 0 1 2 3 4
# 2 4 6 8
# 0: apple
# 1: banana
# 2: cherry
# Tom 18
# Jerry 20

# 在这里写你的代码
for i in range(5):
    print(i)
for i in range(2,10,2):
    print(i)

fruits=["apple", "banana", "cherry"]
for i,k in enumerate(fruits):
    print(f"{i}:{k}")

names=["Tom", "Jerry"]
ages=[18, 20]
for k,v in zip(names,ages):
    print(f"{k} {v}")
```

### 5、while 循环

```python
# 1. 用 while 打印 3 2 1（n 从 3 递减到 1）
# 2. 用 while True + break 实现：count 从 0 开始累加，到 3 时停止
# 期望输出：
# 3
# 2
# 1
# 1
# 2
# 3

# 在这里写你的代码
n = 3
while n > 0:
    print(n)
    n=n-1

count = 1
while True:
    count = count + 1
    print(count)
    if count == 3:
        break
print(count)
```

### 6、break / continue

```python
# 遍历 range(10)，用 continue 跳过 3，用 break 在 6 时停止
# 期望输出：0 1 2 4 5

# 在这里写你的代码
for i in range(10):
    if i == 3:
        continue
    if i == 6:
        break
    print(i)
```

### 7、for-else / while-else 查找

```python
# 1. 用 for-else 遍历 nums=[2,4,6,8] 查找 target=5，没找到打印 "没找到"
# 2. 用 for-else 遍历 nums=[2,4,6,8] 查找 target=6，找到了打印 "找到了" 并 break
# 期望输出：
# 没找到
# 找到了

# 在这里写你的代码
nums=[2,4,6,8]
target = 5
for i in nums:
    if i == target:
        break
else:
    print("没找到")
nums=[2,4,6,8]
target = 6
for i in nums:
    if i == target:
        print("找到了")
        break
else:
    print("没找到")
```

### 8、真值测试（Truthy / Falsy）

```python
# 1. 用 if lst: 判断 [1,2,3] 是否为空
# 2. 用 if not lst: 判断 [] 是否为空
# 3. 用 if s: 判断 "hello" 是否为空字符串
# 4. 用 if x: 判断 x=0 是否为零
# 5. 用 if val is None: 判断 val=None
# 期望输出：
# 列表非空
# 列表为空
# 字符串非空
# x 是零
# val 是 None

# 在这里写你的代码
lst = [1,2,3]
if lst:
    print("列表非空")
lst = []
if not lst:
    print("非空")
s = "hello"
if s:
    print("非空字符串")
x = 0
if x:
    print("x不等于0")

val = None
if val is None:
    print("val是None")
```
