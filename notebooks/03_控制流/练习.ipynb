### 1、if / elif / else 条件分支

```python
# 写一个判断：输入分数 score = 73
# >=90 是 "A"，>=80 是 "B"，>=60 是 "C"，否则是 "D"，打印等级
# 期望输出：C

# 在这里写你的代码
def score(score:int):
    if score >= 90:
        print("A")
    elif score >= 80:
        print("B")
    elif score >= 60:
        print("C")

score(73)
```
